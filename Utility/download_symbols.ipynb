```python
import requests
import time
from datetime import datetime, timedelta

start_date = datetime(2020, 7, 13)
end_date = datetime(2026, 8, 12)

current_date = start_date
while current_date <= end_date:
    date_str = current_date.strftime('%Y-%m-%d')
    url = f"https://api.edinet-fsa.go.jp/api/v2/documents.json?date={date_str}&type=2&Subscription-Key={api_key}"


    try:
        res = requests.get(url, timeout=600000)
    except ConnectionError:
        continue
    except ConnectionError:
        continue
    if res.status_code != 200:
        print(f"Error: {res.status_code} on {date_str}")
        current_date += timedelta(days=1)
        continue

    data = res.json()

    for item in data['results']:
        # exclude RIET etc
        if item["fundCode"] is None:
            # excluded no info record
            if item['submitDateTime'] is not None:
                title = item['filerName']
                if title is None:
                    title = ""
                else:
                    title += " - "
                description = item['docDescription']
                if description:
                    title += description
                doc = {
                    'date': item['submitDateTime'],
                    'documentId': item['docID'],
                    'type': item['docTypeCode'],
                    'title': title,
                    'formCode': item['formCode'],
                    'ordinanceCode': item['ordinanceCode'],
                    'disclosureStatus': item['disclosureStatus'],
                    'periodStart': item['periodStart'],
                    'periodEnd': item['periodEnd'],
                    'edinetCode': item['edinetCode'],
                    'secCode': item['secCode']
                }

                edinet_code = doc['edinetCode']
                cur.execute("SELECT id FROM companies WHERE edinet_code = ?", (edinet_code,))
                company_id = cur.fetchone()
                if company_id is not None:
                    company_id = company_id[0]
                    cur.execute("""
                        INSERT INTO edinet_documents (company_id, document_id, date, type, title, formCode, ordinance_code, disclosure_status, period_start, period_end)
                        VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?, ?)
                        ON CONFLICT(document_id) DO UPDATE SET
                            date = excluded.date,
                            type = excluded.type,
                            title = excluded.title,
                            formCode = excluded.formCode,
                            ordinance_code = excluded.ordinance_code,
                            disclosure_status = excluded.disclosure_status,
                            period_start = excluded.period_start,
                            period_end = excluded.period_end
                        """, (company_id, doc['documentId'], doc['date'], doc['type'], doc['title'], doc['formCode'], doc['ordinanceCode'], doc['disclosureStatus'],
                            doc['periodStart'], doc['periodEnd']))

                    conn.commit()

    time.sleep(1)
    current_date += timedelta(days=1)
cur.close()
```

```python
from lxml import html
import glob
import pandas as pd

doc_id_folders = glob.glob(f"{doc_type}/*")
tag_df = pd.DataFrame()
doc_dict = {}
for doc_id_folder in doc_id_folders:
    doc_id = os.path.basename(doc_id_folder)

    files = glob.glob(f"{doc_type}/{doc_id}/*.html")
    tag_represantation = {}
    parent = ""
    child_added = False
    for file in files:
        # initiate key and value for file
        tag = None
        content_rep = None

        with open(file, mode="r", encoding="utf-8") as fp:
            content = fp.read()
        doc = html.fromstring(content)
        tag = doc.xpath('//title')
        if len(tag) == 0:
            print(f"tag not found for a file: {file}")
            continue
        tag = tag[0].text_content()
        # print("---------")
        # print(file)
        # print(tag[0].text_content())

        h_elements = doc.xpath('//h2')
        if len(h_elements) > 0:
            element = h_elements[0]
            content_rep = element.text_content()
            tag_represantation[tag] = content_rep
            parent = content_rep
            child_added = False
            continue

        h_elements = doc.xpath('//h3')
        if len(h_elements) > 0:
            element = h_elements[0]
            content_rep = element.text_content()
            tag_represantation[tag] = content_rep
            if child_added:
                parent = content_rep
            elif len(parent) > 0:
                parent = f"{parent}>{content_rep}"
            else:
                parent = content_rep
            child_added = True
            continue

        h_elements = doc.xpath('//h4')
        if len(h_elements) > 0:
            element = h_elements[0]
            content_rep = element.text_content()
            tag_represantation[tag] = f"{parent}>{content_rep}"
            child_added = True
            continue

        h_elements = doc.xpath('//h5')
        if len(h_elements) > 0:
            element = h_elements[0]
            content_rep = element.text_content()
            tag_represantation[tag] = f"{parent}>{content_rep}"
            child_added = True
            continue
    
    file_tag_df = pd.DataFrame(list(tag_represantation.values()), index=tag_represantation.keys())
    doc_dict[doc_id] = file_tag_df
```

```python
import re

def extract_bracketed(texts):
    # print(texts)
    converted_texts = []
    for text in texts:
        if isinstance(text, str):
            if "【" in text:
                extracted_texts = re.findall(r'【(.*?)】', text)
                if len(extracted_texts) == 1:
                    converted_texts.append(extracted_texts[0])
                else:
                    converted_texts.append(">".join(extracted_texts))
            else:
                converted_texts.append(text)
        else:
            converted_texts.append(text)
    return converted_texts
```

```python
def html_tables_to_tsvs(table_contents):
    tsv_lines = []
    rows = table_contents.xpath('.//tr')

    for row in rows:
        cells = row.xpath('./th|./td')
        cell_texts = [cell.text_content().strip().replace('\t', ' ').replace('\n', '') for cell in cells]
        cell_text = '\t'.join(cell_texts)
        if len(cell_text) > len(cell_texts):
            tsv_lines.append(cell_text)

    return '\n'.join(tsv_lines)
```

```python
def html_tables_to_md(table_contents):
    data = []
    rows = table_contents.xpath('.//tr')

    for row in rows:
        cells = row.xpath('./th|./td')
        cell_texts = [cell.text_content().strip().replace('|', '\\|').replace('\n', '') for cell in cells]
        cell_text = '_'.join(cell_texts)
        if len(cell_text) > len(cell_texts):
            data.append(cell_texts)
    
    md_lines = []
    if len(data) > 0:
        header = data[0]
        md_lines.append('| ' + ' | '.join(header) + ' |')
        md_lines.append('| ' + ' | '.join(['---'] * len(header)) + ' |')

        for row in data[1:]:
            row = row + [''] * (len(header) - len(row))
            md_lines.append('| ' + ' | '.join(row) + ' |')

        return '\n'.join(md_lines) + '\n'
    return '\n'
```

```python
import re

import pandas as pd

def get_colspan(cell):
    if 'colspan' in cell.attrib:
        colspan = int(cell.attrib['colspan'])
    else:
        colspan = 1
    return colspan

def get_rowspan(cell):
    if 'rowspan' in cell.attrib:
        rowspan = int(cell.attrib['rowspan'])
    else:
        rowspan = 1
    return rowspan

def extract_text_with_br(element):
    texts = []
    if element.text:
        texts.append(element.text)

    for node in element.iterdescendants():
        if node.tag == 'br':
            texts.append('\n')
        elif node.text:
            texts.append(node.text)
        if node.tail:
            texts.append(node.tail)
        texts.append('\n')
    return ''.join(texts[:-1]).strip()


def split_with_strong_newline(text):
    chunks = re.split(r'\n{2,}', text.strip())
    chunks = [chunk.replace("\n", "") for chunk in chunks]
    return chunks

def split_with_special_char(text):
    open_to_close = {
        '(': ')',
        '（': '）',
        '[': ']',
        '【': '】',
    }
    open_brackets = ''.join(open_to_close.keys())
    pattern = rf"""
        [^{open_brackets}]+?
        |
        (
            [{open_brackets}]
            [^)]*?
            [\])）】]
        )
    """
    matches = re.findall(pattern, text, flags=re.VERBOSE)
    
    return [m.strip() for m in matches if m.strip()]

def split_with_length(text, rowspan):
    weak_chunks = re.split(r'\n+', text.strip())
    mid = len(weak_chunks) // rowspan

    text_lines = []
    for i in range(1, rowspan):
        element = ' '.join(weak_chunks[(i-1) * mid: i*mid])
        text_lines.append(element)
    last = ' '.join(weak_chunks[(rowspan -1) * mid:])
    text_lines.append(last)
    return text_lines

def get_rowspan_info(cell, row_index, rowspan, column_index):
    text = extract_text_with_br(cell)
    colspan = get_colspan(cell)
    text_lines = []
    for candidate in text.split("\n"):
        candidate = candidate.strip()
        if len(candidate) > 0:
            text_lines.append(candidate)
    
    info_ele = {
        "row_index": row_index,
        "col_span": colspan,
        "row_span": rowspan,
        "column_index": column_index,
    }
    if len(text_lines) == rowspan:
        info_ele.update({
            "content": text_lines,
            "type": "text_line"
        })
        text = text_lines[0]
    elif len(text_lines) > rowspan:
        text_lines = split_with_strong_newline(text)
        if len(text_lines) == rowspan:
            info_ele.update({
                "content": text_lines,
                "type": "text_line"
            })
            text = text_lines[0]
        else:
            text_lines = split_with_special_char(text)
            if len(text_lines) == rowspan:
                info_ele.update({
                    "content": text_lines,
                    "type": "text_line"
                })
                text = text_lines[0]
            else:
                text_lines = split_with_length(text, rowspan)
                info_ele.update({
                    "content": text_lines,
                    "type": "text_line"
                })
                text = text_lines[0]
    else:
        info_ele.update({
            "content": text,
            "type": "text"
        })
    return text, info_ele

def get_text_from_colspan_element(info_ele, row_index):
    if info_ele["type"] == "text_line":
        contents = info_ele["content"]
        last_span_index = info_ele["row_index"] + info_ele["row_span"] - 1
        line_index = len(contents) - (last_span_index - row_index) -1
        return contents[line_index]
    else:
        return info_ele["content"]

def handle_colspan(cells, cell_text, hspan, colspan, column_index, row_index, header_index, possible_row_spans, rowspan_info):
    total_span = colspan
    text = cell_text
    while hspan > total_span:
        # is parent_text exists, use it as parent. Otherwise, obtain next cell as child 
        cell_hspan = 0
        if len(possible_row_spans) > 0:
            possible_column_indices = [info_ele["column_index"] for info_ele in possible_row_spans]
            if column_index in possible_column_indices:
                for info_ele in possible_row_spans:
                    if info_ele["column_index"] == column_index:
                        span_text = get_text_from_colspan_element(info_ele, row_index)
                        cell_hspan += info_ele["col_span"]
                        if column_index == header_index:
                            text = span_text + " " + text
                        else:
                            text += "" + span_text
            elif (column_index + 1) in possible_column_indices:
                for info_ele in possible_row_spans:
                    if info_ele["column_index"] == column_index:
                        cell_hspan += info_ele["col_span"]
                        span_text = get_text_from_colspan_element(info_ele, row_index)
                        text += " " + span_text
            else:
                cell = next(cells)
                cell_hspan += get_colspan(cell)
                text += " " + cell.text_content().strip().replace('\n', '')
        else:
            cell = next(cells)
            cell_hspan += get_colspan(cell)
            rowspan = get_rowspan(cell)
            if rowspan > 1:
                column_index += 1
                span_text, info_ele = get_rowspan_info(cell, row_index, rowspan, column_index)
                rowspan_info.append(info_ele)
                text += " " + span_text
            else:
                text += " " + cell.text_content().strip().replace('\n', '')
        column_index += 1
        total_span += cell_hspan
    return text, column_index

def html_tables_to_df(table_contents, ignore_row_indices = None, headers=None, index_col=0):
    data = []
    rows = table_contents.xpath('.//tr[not(@hidden)]')
    if ignore_row_indices is not None:
        rows = [row for i, row in enumerate(rows) if i not in ignore_row_indices]

    required_length = 1
    if headers is not None:
        required_length = 0

    if len(rows) > required_length:
        columns = []
        row_initial_index = 0
        headers_span = []
        header_row_spans = []

        if headers is None:
            # temporary define total span for checking header. Will be updated later
            total_span = sum(get_colspan(cell) for cell in rows[0].xpath("./th|./td"))
            for i in range(len(rows)):
                header_cells = rows[i].xpath("./th|./td")
                header_text = "".join([cell.text_content() for cell in header_cells])
                header_text = header_text.strip().replace('\n', '')
                # Assume header row does not have empty text
                if len(header_text) > 1:
                    row_initial_index = i + 1
                    index_cell = header_cells[0]
                    header_row_span = get_rowspan(index_cell)
                    header_col_span = get_colspan(index_cell)
                    for cell in header_cells[1:]:
                        next_cell_text = cell.text_content().strip().replace('\n', '')
                        if len(next_cell_text) == 0:
                            # header_row_span = max(header_row_span, get_rowspan(cell))
                            header_col_span += get_colspan(cell)
                        else:
                            break
                    # if only two cells and total span is not 2, assume first cell is section
                    if len(header_cells) == 2 and  total_span != 2:
                        if header_row_span == 2:
                            # assume first row is section or something.
                            colspan = get_colspan(index_cell)
                            headers_span.append(colspan)
                            index_text = index_cell.text_content().strip().replace('\n', '')
                            # minus 1 to ignore section row
                            header_row_spans.append({"span": header_row_span-1, "text": index_text})
                            columns.append(index_text)
                            continue
                        if header_row_span == 1:
                            # assume first row is section or something and no clear columns present
                            colspan = get_colspan(index_cell)
                            headers_span.append(colspan)
                            text = index_cell.text_content().strip().replace('\n', '')
                            header_row_spans.append(
                                {"span": header_row_span, "text": text}
                            )
                            columns.append(text)
                            section_cell = header_cells[1]
                            colspan = get_colspan(section_cell)
                            section_txt = section_cell.text_content().strip().replace('\n', '')
                            for i in range(colspan):
                                headers_span.append(1)
                                header_row_spans.append({"span": 1, "text": "_dummy_"})
                                columns.append(f"{section_txt}_{i+1}")
                            break
                        else:
                            raise NotImplementedError("Not Implemented header row span > 2 with section header")
                    # There is 2nd index, but header is section or something
                    elif len(header_cells) == total_span - header_col_span:
                        if header_row_span == 2:
                            sum_colspan = 0
                            for cell in header_cells:
                                colspan = get_colspan(cell)
                                headers_span.append(colspan)
                                text = cell.text_content().strip().replace('\n', '')
                                header_row_spans.append({"span": get_rowspan(cell) - 1, "text": text})
                                columns.append(text)
                                sum_colspan += colspan
                                if sum_colspan >= header_col_span:
                                    break
                        else:
                            raise NotImplementedError("Not Implemented header row span != 2 with section header")
                    else:
                        index_cell = header_cells[0]
                        # typical case with all columns
                        for cell in header_cells:
                            colspan = get_colspan(cell)
                            headers_span.append(colspan)
                            text = cell.text_content().strip().replace('\n', '')
                            header_row_spans.append({"span": get_rowspan(cell), "text": text})
                            columns.append(text)
                        break
        else:
            columns = headers
            header_row_span = 1
            for i in range(len(headers)):
                headers_span.append(1)
                header_row_spans.append({"span": 1, "text": ""})

        rowspan_info = []
        total_span = sum(headers_span)
        for row_index, row in enumerate(rows[row_initial_index:]):
            cells = iter(row.xpath('./th|./td'))
            # if colspan is the same, use text as is. if not, join text within colspan
            header_index = 0
            column_index = 0

            cell_texts = []
            possible_row_spans = []

            index_row_span = header_row_spans[0]
            # if header row span is greater than 1, use index_row_span to fill index column
            if index_row_span["span"] - 1 > row_index:
                # hspan = headers_span[header_index]
                # index column doesn't need to check hspan
                cell_texts.append(index_row_span["text"])
                column_index += 1
                header_index += 1
            
            # if past row span is greater than 1, use it to fill cell_texts. Filter rowspan_info first
            for info_ele in rowspan_info:
                # check if ele.column_index is the same && ele.row_index + ele.row_span >= row_index
                last_span_index = info_ele["row_index"] + info_ele["row_span"] - 1
                if info_ele["column_index"] <= row_index and last_span_index >= row_index:
                    possible_row_spans.append(info_ele)
            while True:
                try:
                    # check filtered possible_row_spans to see if current column_index exists
                    if len(possible_row_spans) > 0 and len(headers_span) > header_index:
                        hspan = headers_span[header_index]
                        possible_column_indices = [info_ele["column_index"] for info_ele in possible_row_spans]
                        while column_index in possible_column_indices:
                            for info_ele in possible_row_spans:
                                if info_ele["column_index"] == column_index:
                                    if info_ele["type"] == "text_line":
                                        contents = info_ele["content"]
                                        last_span_index = info_ele["row_index"] + info_ele["row_span"] - 1
                                        line_index = len(contents) - (last_span_index - row_index) -1
                                        span_cell_text = contents[line_index]
                                        column_index += 1
                                        if hspan == 1:
                                            cell_texts.append(span_cell_text)
                                        else:
                                            pre_text, column_index = handle_colspan(cells, span_cell_text, hspan, info_ele["col_span"], column_index, row_index, header_index, possible_row_spans, rowspan_info)
                                            cell_texts.append(pre_text)
                                    else:
                                        column_index += 1
                                        if hspan == 1:
                                            cell_texts.append(info_ele["content"])
                                        else:
                                            pre_text, column_index = handle_colspan(cells, info_ele["content"], hspan, info_ele["col_span"], column_index, row_index, header_index, possible_row_spans, rowspan_info)
                                            cell_texts.append(pre_text)
                                    header_index += 1
                                    if header_index < len(headers_span):
                                        hspan = headers_span[header_index]
                                    else:
                                        break
                    cell = next(cells)
                    colspan = get_colspan(cell)
                    if colspan == total_span:
                        # print(f"Unusual colspan {colspan} at row {row_index} col {column_index} with header {header_index}: {cell.text_content().strip()}")
                        print("ignore cell with full colspan")
                        continue
                    rowspan = get_rowspan(cell)
                    if header_index < len(headers_span):
                        hspan = headers_span[header_index]
                    else:
                        break

                    if rowspan > 1:
                        text, info_ele = get_rowspan_info(cell, row_index, rowspan, column_index)
                        rowspan_info.append(info_ele)
                    else:
                        text = cell.text_content().strip().replace('\n', '')
                                    
                    if hspan > colspan:
                        # print(f"Ref : {hspan} > {colspan}, at row {row_index} col {column_index} with header {header_index}")
                        text, column_index = handle_colspan(cells, text, hspan, colspan, column_index, row_index, header_index, possible_row_spans, rowspan_info)
                        cell_texts.append(text)
                    elif hspan < colspan:
                        print(f"Not Implemented: {hspan} < {colspan}, at row {row_index} col {column_index} with header {header_index}")
                        # raise NotImplementedError("Column span handling not implemented")
                    else:
                        cell_texts.append(text)
                        column_index += 1
                    header_index += 1
                except StopIteration:
                    break
                except NotImplementedError as nie:
                    raise nie
                    continue
                except Exception as e:
                    print(f"error at row {row_index} col {column_index} with header {header_index}")
                    raise  e
                    continue
            if len(cell_texts) > 0:
                data.append(cell_texts)
        
        md_lines = {}
        if len(data) > 0:
            if index_col is not None and index_col >= 0:
                for row in data:
                    row_key = row[index_col]
                    md_lines[row_key] = [*row[:index_col], *row[index_col+1:]]
                df = pd.DataFrame.from_dict(md_lines, orient="index", columns=columns[1:])
                df.index.name = columns[0]
            else:
                for index, row in enumerate(data):
                    md_lines[index] = row
                df = pd.DataFrame.from_dict(md_lines, orient="index", columns=columns)
            return df

html_tables_to_df(table)
```

```python
import calendar
import datetime

import numpy as np

NUMBERS = '0123456789'

KANJI_NUM_MAP = {
    '〇': 0, '零': 0,
    '一': 1, '壱': 1, '弌': 1,
    '二': 2, '弐': 2, '貳': 2,
    '三': 3, '参': 3, '參': 3,
    '四': 4,
    '五': 5,
    '六': 6,
    '七': 7,
    '八': 8,
    '九': 9
}

KANJI_UNIT_MAP = {
    '銭': 0.01,
    '十': 10, '拾': 10,
    '百': 100, '佰': 100,
    '千': 1000, '仟': 1000,
    '万': 10**4, '萬': 10**4,
    '億': 10**8, 
    '兆': 10**12,
}

def kanji_to_int(text):
    result = 0
    section = 0  # e.g., 1234 in "一億二千三百四十五万"
    num = 0  # currently accumulating number

    for c in text:
        if c in KANJI_NUM_MAP:
            num = KANJI_NUM_MAP[c]
        elif c in NUMBERS:
            num = num*10 + int(c)
        elif c in KANJI_UNIT_MAP:
            unit = KANJI_UNIT_MAP[c]
            if unit >= 10000:
                section = (section + num) * unit
                result += section
                section = 0
            else:
                if num == 0:
                    num = 1
                section += num * unit
            num = 0
        else:
            continue

    result += section + num
    return result

def replace_zenkaku(text):
    return text.translate(str.maketrans('０１２３４５６７８９', '0123456789'))

def replace_number(text):
    return kanji_to_int(replace_zenkaku(text))

def convert_to_month_end(date_str):
    date_str = replace_zenkaku(date_str)
    if date_str.startswith("平成"):
        match = re.match(r"平成(.*?)年", date_str)
        jp_year_str = match[1]
        if jp_year_str == "元":
            jp_year = 1
        else:
            jp_year = int(jp_year_str)
        year = 1988 + jp_year
        date_str = date_str.replace(f"平成{jp_year_str}", f"{year}")
    elif date_str.startswith("令和"):
        match = re.match(r"令和(.*?)年", date_str)
        jp_year_str = match[1]
        if jp_year_str == "元":
            jp_year = 1
        else:
            jp_year = int(jp_year_str)
        year = 2018 + int(jp_year)
        date_str = date_str.replace(f"令和{jp_year_str}", f"{year}")
    if "至" in date_str:
        match = re.search(r"至\s*([0-9]+)年\s*([0-9]+)月", date_str)
        if match:
            year, month = match.groups()
            date_str = f"{year}年{month}月"
        else:
            raise ValueError(f"Cannot parse date string: {date_str}")
    if date_str.endswith("期"):
        date_str = date_str[:-1]
    match = re.search(r"\s*([0-9]+)年\s*([0-9]+)月", date_str)
    if match:
        year, month = match.groups()
        date_str = f"{year}年{month}月"
    format = ""
    if "年" in date_str:
        format += "%Y年"
    if "月" in date_str:
        format += "%m月"
    if "日" in date_str:
        format += "%d日"
    if format == "":
        return None
    dt = datetime.datetime.strptime(date_str, format)
    last_day = calendar.monthrange(dt.year, dt.month)[1]
    return datetime.date(dt.year, dt.month, last_day)

def split_label_and_unit(text):
    match = re.match(r'^(.*?)\s*[（(](.*?)[）)]\s*$', text)
    if match:
        label = match.group(1).strip()
        unit = match.group(2).strip()
        return label, unit
    return text.strip(), None

def is_unit_value(value):
    if pd.isna(value):
        return False
    text = str(value)
    pattern = r"^(\$|€|£|¥|円|％|%|倍|人|件|台|点|kg|g|m|km|cm|時間|日|年|〃)$"
    return bool(re.match(pattern, text))

def is_unit_column(series, threshold=0.8):
    valid = series.dropna().map(is_unit_value)
    if len(valid) == 0:
        return False
    return valid.mean() >= threshold

def split_symbol(text):
    match = re.match(r'^(.*?)([¥\\$€円％%倍人]+)$', text)
    if match:
        label = match.group(1).strip()
        symbol = match.group(2).strip()
        if symbol in ["円", "¥"]:
            symbol = "JPY"
        elif symbol in ["$"]:
            symbol = "USD"
        elif symbol in ["€"]:
            symbol = "EUR"
        elif symbol in ["%", "％"]:
            symbol = "PCT"
        elif symbol in ["倍"]:
            symbol = "RTO"
        elif symbol in ["人"]:
            symbol = "NoP"
        return label, symbol
    return text, None

def remove_deliminater(text):
    return text.replace(",", "")


def split_sign(text):
    if text.startswith(('▲', '▼', '△')):
        return -1, text[1:]
    else:
        return 1, text

def convert_with_unit(num_txt: str, unit):
    if num_txt is None or (isinstance(num_txt, float) and np.isnan(num_txt)):
        return np.nan
    num_txt = num_txt.replace("　", "").replace("\xa0", "").strip()
    if len(num_txt) == 0:
        return np.nan
    match = re.search(r'[〔（(［\[](.*?)[］\])）〕]', num_txt)
    if match:
        num_txt = match.group(1).strip()
    
    if "-" == num_txt or "－" == num_txt or num_txt == "―" or num_txt == "ー":
        return np.nan
    num_txt, _ = split_symbol(num_txt)
    num_txt = remove_deliminater(num_txt)
    sign, num_txt = split_sign(num_txt) 
    if "." in num_txt:
        number = float(num_txt)
        unit = float(unit)
    else:
        number = kanji_to_int(num_txt)
    number *= sign * unit
    return number
```
